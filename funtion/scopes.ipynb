### 3가지 함수 유형별 코드 실행 결과 비교
| 구분 | 리턴값 없음<br>(print만 있음) | 리턴값만 있음<br>(return만 있음) | 리턴값+출력값 있음<br>(print + return) |
|:----:|:-------------------------------:|:-----------------------------------:|:--------------------------------------:|
| 1 | **result = print_message()**<br>함수 실행, "메시지 출력!" 출력<br>result = None | **result = print_message()**<br>함수 실행, 출력 없음<br>result = "바보야 이게 리턴 값이야!" | **result = print_message()**<br>함수 실행, "출력 값" 출력<br>result = "바보야 이게 리턴 값이야!" |
| 2 | **func = print_message**<br>함수 객체 할당, 실행X | **func = print_message**<br>함수 객체 할당, 실행X | **func = print_message**<br>함수 객체 할당, 실행X |
| 3 | **print('result : ', result)**<br>"result :  None" 출력 | **print('result : ', result)**<br>"result :  바보야 이게 리턴 값이야!" 출력 | **print('result : ', result)**<br>"result :  바보야 이게 리턴 값이야!" 출력 |
| 4 | **print('func():', func())**<br>func() 실행 → "메시지 출력!" 출력<br>print로 "func(): None" 출력 | **print('func():', func())**<br>func() 실행, 출력 없음<br>print로 "func(): 바보야 이게 리턴 값이야!" 출력 | **print('func():', func())**<br>func() 실행 → "출력 값" 출력<br>print로 "func(): 바보야 이게 리턴 값이야!" 출력 |
| 5 | **print(func)**<br>함수 객체 주소 출력 | **print(func)**<br>함수 객체 주소 출력 | **print(func)**<br>함수 객체 주소 출력 |
| 6 | **print(print_message)**<br>함수 객체 주소 출력 | **print(print_message)**<br>함수 객체 주소 출력 | **print(print_message)**<br>함수 객체 주소 출력 |
| 7 | **func()**<br>함수 실행, "메시지 출력!" 출력 | **func()**<br>함수 실행, 출력 없음 | **func()**<br>함수 실행, "출력 값" 출력 |



```python
def print_message(): #리턴값이 없는 함수
    print("메시지 출력!")

print("1--------")
result = print_message()# 먼저 왼쪽 함수실행(출력),result에는 None 저장(함수는 리턴값이 없으므로)
print()

print("2--------")
func = print_message # 왼쪽의 함수는 실행하지 않는다.func는 print_message 함수 객체
print()

print("3--------")
print('result : ',result)    # None 출력(func는 함수 객체일뿐 아직 함수는 실행하지 않았으므로 none 출력)
print()

print("4--------")
print('func():',func())  #print함수의 인자인 func()는 print_message 함수 객체를 실행한 결과를 출력하고, 그 다음 print함수가 실행된다 그러므로 이때 인자의 반환값은 None이므로 출력결과는 None
print()

print("5--------")
print(func)      # <function print_message at 0x...> 함수 객체의 메모리 주소 출력
print()

print("6--------")
print(print_message) # <function print_message at 0x...> 함수 객체의 메모리 주소 출력
print()

print("7--------")
func()# print_message 함수 객체를 실행

print("8--------")
#result()# 1을 실행했을때 result는 none을 갇는다. 즉 result는 함수 객체가 아니므로 에러가 발생한다.
```

함수의 실행이 끝나면 함수안에 있는 함수, 변수는 사라진다. (Garbage Collector에 의해서)

```python
name = '크래프트맨'
def outer():
    name = '멘탈리티'
    print(name)
    def inner():
        name = '크래프트맨 엔탈리티'
        print('inner()의 id :', name, id(inner))
        print(name)
    inner()


outer()
print('-------')
print('outer()의 id :', id(outer))
print('inner()의 id :', id(inner)) #NameError: name 'inner' is not defined
```

nonlocal keywords(잘 사용하지 않음)

```python
def outer( ):
    name = '크래프트맨'#enclosed변수
    print('outer' , name)
    def inner():
        name =name + '멘탈리티' #local변수로 enclosed변수를 바꿀 수 없다. 에러예상.
        print('inner', name)
    inner()
outer( ) #local variable 'name' referenced before assignment 당연히 에러발생
```

```python
#하지만, 꼭! 바꾸고 싶다면
def outer( ):
    name = '크래프트맨'#enclosed변수
    print('outer' , name)
    def inner():
        nonlocal name #당연히, enclosed변수가 존재해야 함.
        name =name + ' 멘탈리티' 
        print('inner', name)
    inner()
    print('outer after update', name) # enclose변수(name)이 바뀐 것을 알 수 있다. 그래서 위험.
outer( )
```

```python
def outer( ):
    name = '크래프트맨'#enclosed변수
    print('outer' , name)
    def inner():
        name ='멘탈리티' 
        print('inner', name)
    return inner #함수 실행(X),객체만 리턴한다.
    
outer( )
print('-------')
print()

a = outer #outer 함수 객체
print('-------')
print()

b = outer() # outer()를 실행해서, 반환값(inner 함수 객체)을 b에 저장
print('-------')
print()

a() 
print('-------')
print()

b() # inner() 실행 → "inner 멘탈리티" 출력
print('-------')
print()

print(a)
print('-------')
print()

print(b)
```

```python
def outer( ):
    name = '크래프트맨'#enclosed변수
    print('outer' , name)
    def inner():
        nonlocal name
        name =name + ' 천재' # 계속 name이 변한다.
        print('inner', name)
    return inner #이경우에슨 inner함수를 return하지만, outer()에서 실행하지는 않는다.
    
outer( )
print('------')
a = outer()
print('------')
b = outer()
print('------')
a()
b()
print('------')
a() # ->위 a()의 return값을 받아서실행 nonlocal name = 크래프트맨 천재값을 받아 inner()실생
print('------')
a()
print('------')
b() # ->위 a()의 return값을 받아서실행 nonlocal name = 크래프트맨 천재 천재값을 받아 inner()실생
print('------')
outer()
print('------')
```
